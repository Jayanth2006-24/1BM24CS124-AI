```python
def alpha_beta(depth, is_max, alpha, beta, path):
    global index, pruned

    if depth == len(branching):
        value = leaf_values[index]
        index += 1
        return value, []

    if is_max:
        best = float('-inf')
    else:
        best = float('inf')

    best_path = []

    for i in range(branching[depth]):
        if beta <= alpha:
            pruned.append(path + [i + 1])
            skip = 1
            for b in branching[depth + 1:]:
                skip *= b
            index += skip
            continue

        value, sub_path = alpha_beta(
            depth + 1, not is_max, alpha, beta, path + [i + 1]
        )

        if is_max:
            if value > best:
                best = value
                best_path = [i + 1] + sub_path
            alpha = max(alpha, best)
        else:
            if value < best:
                best = value
                best_path = [i + 1] + sub_path
            beta = min(beta, best)

    return best, best_path


levels = int(input("Enter number of levels: "))
branching = []

for i in range(levels):
    branching.append(int(input(f"Enter branching factor for level {i + 1}: ")))

leaf_count = 1
for b in branching:
    leaf_count *= b

leaf_values = list(map(int, input(f"Enter {leaf_count} leaf values: ").split()))

if len(leaf_values) != leaf_count:
    print("Invalid number of leaf values")
else:
    index = 0
    pruned = []

    root_value, best_path = alpha_beta(
        0, True, float('-inf'), float('inf'), []
    )

    print("\nRoot node value:", root_value)
    print("Best path:", " -> ".join(map(str, best_path)))

    if pruned:
        print("Pruned paths:")
        for p in pruned:
            print(" -> ".join(map(str, p)))
    else:
        print("No nodes were pruned")
```
